```python
import localcider
from localcider.sequenceParameters import SequenceParameters
import pandas as pd
```

```python
def kappa(sequence):
    
# create a SequenceParameters object with your amino acid sequence
    SeqOb = SequenceParameters(sequence)

    # carry out analysis
    return (SeqOb.get_kappa())
```

```python
tf1 = pd.read_csv('BP1_properties.csv')
tf2 = pd.read_csv('BP2_properties.csv')
```

```python
tf1['Kappa'] = tf1['BP Sequence'].apply(kappa)
tf2['Kappa'] = tf2['BP Sequence'].apply(kappa)
```
