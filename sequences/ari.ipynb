```python
from sklearn.metrics import adjusted_rand_score
from sklearn.metrics import fowlkes_mallows_score
import pandas as pd
```

```python
df1= pd.read_csv('/project/zerze/edayalam/primitive_clustering/combined_clusters/combined_bp1.csv')
df1
```

```python
df2= pd.read_csv('/project/zerze/edayalam/primitive_clustering/combined_clusters/combined_bp2.csv')
df2
```

```python
df1.loc[1]
```

```python
df = pd.merge(df1, df2, on='UniProt ID', how='inner')
df
```

```python
l1,l2= df['Cluster_x'], df['Cluster_y']
```

```python
print (adjusted_rand_score(l1,l2))
```

```python
print (fowlkes_mallows_score(l1,l2))
```
